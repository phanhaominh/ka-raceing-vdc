# Stage 2: Robustness Analysis

Loads the merged Stage 2 dataset and extracts the speed-consistency trade-off.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('results/robustness/all_results.csv')
samples = np.load('results/robustness/robust_samples.npy')

params = ['yaw_gain','yaw_damping','tc_slip_target','tc_aggressiveness','regen_ratio','brake_bias_front']
for i, name in enumerate(params):
    df[f'param_{name}'] = df['sample_id'].apply(lambda idx: samples[int(idx), i])

print(f'Loaded {len(df):,} samples')
print(f'J range: {df["J"].min():.3f} — {df["J"].max():.3f}')

# Yaw damping sweep
print('\nYaw damping effect on lap time and consistency:')
for d in range(0, 61, 10):
    s = df[(df['param_yaw_damping'] >= d) & (df['param_yaw_damping'] < d+10)]
    if len(s) > 0:
        print(f'  yaw_d={d}-{d+9}: lap={s["mean_lap"].mean():.3f}s  std={s["std_lap"].mean():.4f}s  n={len(s)}')
```
